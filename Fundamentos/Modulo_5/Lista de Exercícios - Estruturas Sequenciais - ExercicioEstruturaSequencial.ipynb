#### 8. Vamos criar um conversor de temperatura. Faça um Programa que peça a temperatura em graus Fahrenheit, transforme e mostre a temperatura em graus Celsius.
$C = \frac{5}{9}(F-32)$

```python
farenheit= float(input('Digite a temperatura em Fahrenheit: '))
celsius= (farenheit - 32) * 5 / 9
print(f'A temperatura digitada em Celsius é: {celsius:.2f} °C')
```

#### 9. Faça um Programa que peça a temperatura em graus Celsius, transforme e mostre em graus Fahrenheit.
$F = \frac{9}{5}C + 32$

```python
celsius= float(input('Digite a temperatura em Celsius: '))
f= (celsius * 9 / 5) + 32
print(f'A temperatura digitada em Fahrenheit é: {f:.2f} °F')
```

#### 11. Tendo como dado de entrada a altura (h) de uma pessoa, construa um algoritmo que calcule seu peso ideal, utilizando as seguintes fórmulas:
##### a. Para homens: $P = 72,7h - 58$
##### b. Para mulheres: $P = 62,1h - 44,7$

```python
sexo= str(input('Digite o sexo (M/F): ')).upper()
h= float(input('Digite o valor da altura: '))
if sexo == 'M':
    p= 72.7 * h - 58
    print(f'O peso ideal para a altura digitada de homens é: {p:.2f} kg')
elif sexo == 'F':
    p= 62.1 * h - 44.7
    print(f'O peso ideal para a altura digitada de mulheres é: {p:.2f} kg')
else:
    print('Sexo inválido. Por favor, digite M para masculino ou F para feminino.')
```

#### 12. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês.

```python
hora= float(input('Quanto você ganha por hora? '))
trab= float(input('Quantas horas você trabalha por mês? '))
```

#####  Calcule o salário bruto (horas * salario por hora)

```python
sal_bruto= hora * trab
```

##### Calcule o desconto do IR (11% do salário bruto)

```python
ir= sal_bruto - (sal_bruto * 11/ 100)
```

##### Calcule o desconto do INSS (8% do salário bruto)

```python
inss= sal_bruto - (sal_bruto * 8/ 100)
```

##### Calcule o desconto do sindicato (5% do salário bruto)

```python
descon_sindicato= sal_bruto - (sal_bruto * 5/ 100)
```

##### Calcule o salário líquido (salário bruto - descontos)

```python
sal_liquido= sal_bruto - ir - inss - descon_sindicato
```

#### 13. Faça um programa para uma loja de tintas. O programa deverá pedir o tamanho em metros quadrados da área a ser pintada. Considere que a cobertura da tinta é de 1 litro para cada 3 metros quadrados e que a tinta é vendida em latas de 18 litros, que custam R\$ 80,00. Informe ao usuário a quantidades de latas de tinta a serem compradas e o preço total. (para simplificação nesse momento, não se preocupe em arredondar a quantidade de latas a serem compradas - vamos trabalhar isso em breve)

```python
area= float(input('Digite a área em m²: '))
litros= area / 3
latas= litros / 18
tot= latas * 80
print(f'Você precisará de {latas:.2f} latas de tinta.')
print(f'O valor total será de R$ {tot:.2f}.')
```

#### 14. Faça um programa que peça o tamanho de um arquivo para download (em MB) e a velocidade de um link de Internet (em Mbps), calcule e informe o tempo aproximado de download do arquivo usando este link (em minutos).

Detalhe: MB significa megabyte, Mb (com b minúsculo) significa megabit. Um megabit é 1/8 de um megabyte. 

```python
Mb= float(input('Digite o valor do download em MB: '))
velocidade= float(input('Digite a velocidade da internet em Mbps: '))
tempo= Mb * 8

velo= tempo / velocidade

tempo_minutos= velo / 60
print(f'O tempo estimado para o download é de {tempo_minutos:.2f} minutos.')
#as
```
